```python

from abc import ABC, abstractmethod


# Abstract Base Class
class BuildingSystem(ABC):
    
    @abstractmethod
    def start(self):
        """Start or activate the system."""
        pass

    @abstractmethod
    def stop(self):
        """Stop or deactivate the 

system."""
        pass

    @abstractmethod
    def status(self):
        """Display the current state of the system."""
        pass
```

```python

# Child Class 1: Air Conditioning System
class AirConditioningSystem(BuildingSystem):
    def __init__(self, location: str = 

"Auditorium Main Hall"):
        self.location = location
        self.is_running = False

    def start(self):
        self.is_running = True
        print(f"[AirConditioningSystem] Unit at '{self.location}' started cooling.")

    def stop(self):
        self.is_running = False
        print(f"[AirConditioningSystem] Unit at '{self.location}' stopped.")


    def status(self):
        state = "Running (Cooling)" if self.is_running else "Off"
        print(f"[AirConditioningSystem] Status: {state}")
```

```python
# Child Class 2: Lighting System
class LightingSystem(BuildingSystem):
    def __init__(self, zone: str = "Stage & Main Seating"):
        self.zone = zone
        self.is_on = False


    def start(self):
        self.is_on = True
        print(f"[LightingSystem] Lights in '{self.zone}' turned ON.")

    def stop(self):
        self.is_on = False
        print(f"[LightingSystem] Lights in '{self.zone}' turned OFF.")

    def status(self):
        state = "ON" if self.is_on else "OFF"
        print(f"[LightingSystem] Status: {state}")
```

```python
# Child Class 3: Security System
class SecuritySystem(BuildingSystem):
    def __init__(self, perimeter: str = "Auditorium Perimeter"):
        self.perimeter = perimeter
        self.is_armed = False

    def start(self):
        self.is_armed = True
        print(f"[SecuritySystem] Armed perimeter monitoring for '{self.perimeter}'.")


    def stop(self):
        self.is_armed = False
        print(f"[SecuritySystem] Disarmed perimeter monitoring for '{self.perimeter}'.")

    def status(self):
        state = "ARMED" if self.is_armed else "DISARMED"
        print(f"[SecuritySystem] Status: {state}")
```

```python
# New Child Class: Fire Alarm System

# Demonstrates extensibility without modifying existing classes
class FireAlarmSystem(BuildingSystem):
    def __init__(self, zone: str = "Entire Auditorium Building"):
        self.zone = zone
        self.is_active = False

    def start(self):
        self.is_active = True
        print(f"[FireAlarmSystem] Smoke and fire detection ACTIVE for '{self.zone}'.")

    def stop(self):
        self.is_active = False
        print(f"[FireAlarmSystem] Fire detection system STANDBY for '{self.zone}'.")

    def status(self):
        state = "MONITORING (Active)" if self.is_active else "STANDBY (Inactive)"
        print(f"[FireAlarmSystem] Status: {state}")
```

```python
systems = [AirConditioningSystem(),LightingSystem(),SecuritySystem(),FireAlarmSystem()]
```

```python
print("Starting all systems....")
for sys in systems:
    sys.start()
```

```python
print("Stoping all systems....")
for sys in systems:
    sys.stop()
```

```python
print("Final Status Check....")
for sys in systems :
    sys.status()
```
